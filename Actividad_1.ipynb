# **2.- Programa en Python que lea 2 números, realizar la suma y desplegar la suma de los 2 números**

```python
n1=int(input("Escribe un numero: "))
n2=int(input("Escribe otro numero: "))
suma=n1+n2
print("La suma de los numeros es=",suma)
```

# **3.- Programa en Python que lea 2 números y realice las 4 operaciones básicas**

```python
n1=int(input("Dame un numero:"))
n2=int(input("Dame otro numero"))
s=n1+n2
r=n1-n2
m=n1*n2
d=n1/n2
print("La suma de los numeros es=",s)
print("La resta de los numeros es=",r)
print("La multiplicacion de los numeros es=",m)
print("La division de los numeros es=",d)
```

# **4.- Programa en Python Que sirva para calcular el área de un triángulo**

```python
base=int(input("Escribe la base="))
alt=int(input("Escribe la altura="))
area=(base*alt)/2
print(f"El area del triangulo es={area}")
```
